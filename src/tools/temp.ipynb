```python
import json
with open('bitwarden_marc.json', 'r') as file:
    data = json.load(file)
```

```python
small = [{'name': item["name"], 'folderId': item['folderId'], 'username': item['login']['username'] if 'login' in item.keys() else ''} for item in data["items"] if item["type"]==1 and 'folderId' in item]
byFolder = {folder['name']: [{'name': item['name'], 'user': item['username']} for item in small if item['folderId']==folder['id'] ]  for folder in data['folders']} | {'no folder': [item['name'] for item in small if item['folderId']=='']}
for key in byFolder.keys():
    print(key)
    for item in byFolder[key]:
        if item['user']!='marc.paelinck@gmail.com':
            print('    '+ json.dumps(item))

```
